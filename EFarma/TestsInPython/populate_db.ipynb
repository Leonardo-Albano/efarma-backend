```python
import requests
from requests.packages.urllib3.exceptions import InsecureRequestWarning
requests.packages.urllib3.disable_warnings(InsecureRequestWarning)
import hashlib
from datetime import datetime

base_url = "https://localhost:7085/api/"
```

```python
url = base_url + "Page"

objects = [
    {"name": "Home", "description": "Tela inicial"},
    {"name": "Cadastros", "description": "Tela de cadastro de pessoas"},
    {"name": "Receitas", "description": "Tela de criação e acesso de receitas"},
    {"name": "Estoque", "description": "Tela de controle do estoque"},
    {"name": "Configurações", "description": "Tela de configurações do administrador"}
]

for obj in objects:
    response = requests.post(url, json=obj, verify=False)
    if response.status_code == 200 or response.status_code == 201:
        print(f"Página '{obj['name']}' criada com sucesso!")
    else:
        print(f"Erro ao criar a página '{obj['name']}': {response.status_code}, {response.text}")
    print()

print("Processo de criação de páginas finalizado.")
```
